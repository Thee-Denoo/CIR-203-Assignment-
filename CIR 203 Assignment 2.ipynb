```python
# Linked list implemented using dictionaries
class StudentLinkedList:
    def __init__(self):
        self.head = None   # Start of linked list

    # Create a dictionary node
    def _create_node(self, name, adm_no, unit1, mark1, unit2, mark2, unit3, mark3, next_node=None):
        return {
            "name": name,
            "admission_number": adm_no,
            "units": [unit1, unit2, unit3],
            "marks": [mark1, mark2, mark3],
            "next": next_node
        }

    # Add student at end of list
    def append(self, name, adm_no, unit1, mark1, unit2, mark2, unit3, mark3):
        new_node = self._create_node(name, adm_no, unit1, mark1, unit2, mark2, unit3, mark3)

        if self.head is None:
            self.head = new_node
            return

        current = self.head
        while current["next"] is not None:
            current = current["next"]
        current["next"] = new_node

    # Print linked list
    def print_list(self):
        current = self.head
        while current is not None:
            print("Name:", current["name"])
            print("Admission No.:", current["admission_number"])
            print("Units:", current["units"])
            print("Marks:", current["marks"])
            print("-----------------------------")
            current = current["next"]

students = StudentLinkedList()
students.append("Alice", "CIT001", "CIR_205", 85, "CIR_203", 90, "CIT_113", 78)
students.append("Bob", "CIT002", "CIR_205", 72, "CIR_203", 88, "CIT_113", 91)
students.append("Charlie", "CIT003", "CIR_205", 95, "CIR_203", 89, "CIT_113", 92)
students.print_list()
```
